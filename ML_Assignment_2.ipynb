```python
from ucimlrepo import fetch_ucirepo
import pandas as pd

# Download Dry Bean dataset
dry_bean = fetch_ucirepo(id=602)

# Separate input features and target
X = dry_bean.data.features
y = dry_bean.data.targets

# Combine them into one DataFrame
df = pd.concat([X, y], axis=1)

# Display basic information
print("Dataset downloaded successfully!")
print("Dataset shape:", df.shape)

print("\nFirst 5 rows:")
display(df.head())
```

```python
# Separate input features
X = df_clean.drop(columns=["Class"])

# Separate target variable
y = df_clean["Class"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)

print("\nFeature columns:")
print(X.columns.tolist())

print("\nTarget classes:")
print(y.unique())
```

```python
from sklearn.model_selection import train_test_split

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display dataset sizes
print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

# Create scaler
scaler = StandardScaler()

# Fit and transform training data
X_train_scaled = scaler.fit_transform(X_train)

# Transform testing data using the same scaler
X_test_scaled = scaler.transform(X_test)

print("Original training shape:", X_train.shape)
print("Scaled training shape:", X_train_scaled.shape)

print("Original testing shape:", X_test.shape)
print("Scaled testing shape:", X_test_scaled.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

# Create Logistic Regression pipeline
logistic_model = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(max_iter=2000, random_state=42))
])

print("Logistic Regression pipeline created successfully!")
```

```python
# Train Logistic Regression model
logistic_model.fit(X_train, y_train)

print("Logistic Regression model trained successfully!")
```

```python
# Predict bean varieties using testing data
y_pred_logistic = logistic_model.predict(X_test)

print("First 10 actual values:")
print(y_test.head(10).tolist())

print("\nFirst 10 predicted values:")
print(y_pred_logistic[:10])
```

```python
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef
)

print("Evaluation metrics imported successfully!")
```

```python
# Calculate evaluation metrics
logistic_accuracy = accuracy_score(y_test, y_pred_logistic)

logistic_auc = roc_auc_score(
    y_test,
    logistic_model.predict_proba(X_test),
    multi_class="ovr",
    average="macro"
)

logistic_precision = precision_score(
    y_test, y_pred_logistic, average="weighted"
)

logistic_recall = recall_score(
    y_test, y_pred_logistic, average="weighted"
)

logistic_f1 = f1_score(
    y_test, y_pred_logistic, average="weighted"
)

logistic_mcc = matthews_corrcoef(y_test, y_pred_logistic)

# Display results
print("LOGISTIC REGRESSION RESULTS")
print("---------------------------")
print("Accuracy:", round(logistic_accuracy, 4))
print("AUC:", round(logistic_auc, 4))
print("Precision:", round(logistic_precision, 4))
print("Recall:", round(logistic_recall, 4))
print("F1 Score:", round(logistic_f1, 4))
print("MCC:", round(logistic_mcc, 4))
```

```python
results = []

results.append({
    "Model": "Logistic Regression",
    "Accuracy": logistic_accuracy,
    "AUC": logistic_auc,
    "Precision": logistic_precision,
    "Recall": logistic_recall,
    "F1 Score": logistic_f1,
    "MCC": logistic_mcc
})

print("Logistic Regression results stored successfully!")
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
# Create the Decision Tree model
decision_tree_model = DecisionTreeClassifier(random_state=42)

# Train the model
decision_tree_model.fit(X_train, y_train)

# Make predictions
y_pred_decision_tree = decision_tree_model.predict(X_test)

print("First 10 actual values:")
print(y_test.head(10).tolist())

print("\nFirst 10 predicted values:")
print(y_pred_decision_tree[:10])
```

```python
# Accuracy
decision_tree_accuracy = accuracy_score(
    y_test, y_pred_decision_tree
)

# AUC Score
decision_tree_auc = roc_auc_score(
    y_test,
    decision_tree_model.predict_proba(X_test),
    multi_class="ovr",
    average="macro"
)

# Precision
decision_tree_precision = precision_score(
    y_test, y_pred_decision_tree, average="weighted"
)

# Recall
decision_tree_recall = recall_score(
    y_test, y_pred_decision_tree, average="weighted"
)

# F1 Score
decision_tree_f1 = f1_score(
    y_test, y_pred_decision_tree, average="weighted"
)

# Matthews Correlation Coefficient
decision_tree_mcc = matthews_corrcoef(
    y_test, y_pred_decision_tree
)
```

```python
results.append({
    "Model": "Decision Tree",
    "Accuracy": decision_tree_accuracy,
    "AUC": decision_tree_auc,
    "Precision": decision_tree_precision,
    "Recall": decision_tree_recall,
    "F1 Score": decision_tree_f1,
    "MCC": decision_tree_mcc
})
```

```python
print("Decision Tree Results")
print("---------------------")
print(f"Accuracy:  {decision_tree_accuracy:.4f}")
print(f"AUC:       {decision_tree_auc:.4f}")
print(f"Precision: {decision_tree_precision:.4f}")
print(f"Recall:    {decision_tree_recall:.4f}")
print(f"F1 Score:  {decision_tree_f1:.4f}")
print(f"MCC:       {decision_tree_mcc:.4f}")
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

knn_model = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", KNeighborsClassifier(n_neighbors=5))
])

print(knn_model)
```

```python
# Train KNN using the training dataset
knn_model.fit(X_train, y_train)

# Predict classes for the test dataset
y_pred_knn = knn_model.predict(X_test)

# Predict class probabilities for AUC calculation
y_prob_knn = knn_model.predict_proba(X_test)

print("KNN predictions completed.")
print("Number of predictions:", len(y_pred_knn))
```

```python
knn_accuracy = accuracy_score(y_test, y_pred_knn)

knn_auc = roc_auc_score(
    y_test,
    y_prob_knn,
    multi_class="ovr",
    average="macro"
)

knn_precision = precision_score(
    y_test, y_pred_knn,
    average="weighted",
    zero_division=0
)

knn_recall = recall_score(
    y_test, y_pred_knn,
    average="weighted",
    zero_division=0
)

knn_f1 = f1_score(
    y_test, y_pred_knn,
    average="weighted",
    zero_division=0
)

knn_mcc = matthews_corrcoef(y_test, y_pred_knn)

print("KNN Model Results")
print(f"Accuracy : {knn_accuracy:.4f}")
print(f"AUC      : {knn_auc:.4f}")
print(f"Precision: {knn_precision:.4f}")
print(f"Recall   : {knn_recall:.4f}")
print(f"F1 Score : {knn_f1:.4f}")
print(f"MCC      : {knn_mcc:.4f}")
```

```python
results.append({
    "Model": "K-Nearest Neighbors",
    "Accuracy": knn_accuracy,
    "AUC": knn_auc,
    "Precision": knn_precision,
    "Recall": knn_recall,
    "F1 Score": knn_f1,
    "MCC": knn_mcc
})

print("KNN results added to comparison.")
```

```python
from sklearn.svm import SVC
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

svm_model = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", SVC(
        kernel="rbf",
        C=1.0,
        probability=True,
        random_state=42
    ))
])

print(svm_model)
```

```python
# Train the SVM model
svm_model.fit(X_train, y_train)

# Predict class labels
y_pred_svm = svm_model.predict(X_test)

# Predict class probabilities for AUC
y_prob_svm = svm_model.predict_proba(X_test)

print("SVM predictions completed.")
print("Number of predictions:", len(y_pred_svm))
```

```python
svm_accuracy = accuracy_score(y_test, y_pred_svm)

svm_auc = roc_auc_score(
    y_test,
    y_prob_svm,
    multi_class="ovr",
    average="macro"
)

svm_precision = precision_score(
    y_test, y_pred_svm,
    average="weighted",
    zero_division=0
)

svm_recall = recall_score(
    y_test, y_pred_svm,
    average="weighted",
    zero_division=0
)

svm_f1 = f1_score(
    y_test, y_pred_svm,
    average="weighted",
    zero_division=0
)

svm_mcc = matthews_corrcoef(y_test, y_pred_svm)

print("SVM Model Results")
print(f"Accuracy : {svm_accuracy:.4f}")
print(f"AUC      : {svm_auc:.4f}")
print(f"Precision: {svm_precision:.4f}")
print(f"Recall   : {svm_recall:.4f}")
print(f"F1 Score : {svm_f1:.4f}")
print(f"MCC      : {svm_mcc:.4f}")
```

```python
results.append({
    "Model": "Support Vector Machine",
    "Accuracy": svm_accuracy,
    "AUC": svm_auc,
    "Precision": svm_precision,
    "Recall": svm_recall,
    "F1 Score": svm_f1,
    "MCC": svm_mcc
})

print("SVM results added to comparison.")
```
